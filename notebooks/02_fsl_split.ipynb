```python
from pathlib import Path
import numpy as np
from sklearn.model_selection import train_test_split

PROJECT_ROOT = Path.cwd().parent
RAW_DIR = PROJECT_ROOT / "datasets" / "raw"
RESULTS_DIR = PROJECT_ROOT / "results"

RESULTS_DIR.mkdir(parents=True, exist_ok=True)

classes = ["normal", "irf", "orf"]

X_list = []
y_list = []
file_list = []

for label, class_name in enumerate(classes):

    X_class = np.loadtxt(
        RAW_DIR / f"{class_name}.csv",
        delimiter=","
    )

    X_list.append(X_class)
    y_list.extend([label] * len(X_class))

    file_list.extend(
        [f"{class_name}_{i:03d}" for i in range(len(X_class))]
    )

X = np.concatenate(X_list, axis=0)
y = np.asarray(y_list)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Class counts:", np.bincount(y))
```

```python
indices = np.arange(len(X))

train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.30,
    stratify=y,
    random_state=42
)

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    stratify=y[temp_idx],
    random_state=42
)

print("Train:", len(train_idx))
print("Validation:", len(val_idx))
print("Test:", len(test_idx))
```

```python
for name, idx in [
    ("TRAIN", train_idx),
    ("VALIDATION", val_idx),
    ("TEST", test_idx)
]:

    print(f"\n{name}")

    counts = np.bincount(y[idx], minlength=3)

    for class_name, count in zip(classes, counts):
        print(f"{class_name}: {count}")
```

```python
np.savez(
    RESULTS_DIR / "fsl_split.npz",
    train_idx=train_idx,
    val_idx=val_idx,
    test_idx=test_idx,
    y=y
)

print("FSL split saved.")
```
